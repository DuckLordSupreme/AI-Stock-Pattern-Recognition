```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def generate_cup(num_points=50):
    x = np.linspace(-1, 1, num_points)

    # Random properties of this particular cup
    depth = np.random.uniform(0.6, 1.4)
    center = np.random.uniform(-0.15, 0.15)
    tilt = np.random.uniform(-0.15, 0.15)
    noise_level = np.random.uniform(0.02, 0.10)

    # Basic cup shape
    cup = depth * (x - center) ** 2

    # Make the left/right sides slightly uneven
    cup = cup + tilt * x

    # Add random noise
    noise = np.random.normal(0, noise_level, num_points)
    cup = cup + noise

    return cup

plt.figure(figsize=(10, 6))
for i in range(10):
    cup = generate_cup()
    plt.plot(cup)

plt.show()
```

```python
def generate_not_cup(num_points=50):
    x = np.linspace(-1, 1, num_points)

    pattern_type = np.random.choice([
        "inverted_cup",
        "v_shape",
        "trend",
        "wave"
    ])

    noise_level = np.random.uniform(0.02, 0.10)

    if pattern_type == "inverted_cup":
        shape = -(x ** 2)

    elif pattern_type == "v_shape":
        shape = np.abs(x)

    elif pattern_type == "trend":
        slope = np.random.uniform(-1, 1)
        shape = slope * x

    elif pattern_type == "wave":
        frequency = np.random.uniform(1, 3)
        shape = np.sin(frequency * np.pi * x)

    noise = np.random.normal(0, noise_level, num_points)

    return shape + noise
```
